```python
import shutil
from pathlib import Path

lang_codes = ["ben", "mar", "tel", "tam"]

src = "./translation/data/method1/clean/final/"
des = "./Translation Experiment/data/langs/"


def move_data():
    src_path = Path(src)
    des_path = Path(des)

    for lang_code in lang_codes:
        source_dir = src_path 
        dest_dir = des_path / lang_code
        dest_file_to_be_deleted = dest_dir / "clean.csv"
        dest_file_to_created = dest_dir / "clean1.csv"

        # 1. Delete existing clean.csv if it exists
        if dest_file_to_be_deleted.exists():
            dest_file_to_be_deleted.unlink()
            print(f"Deleted: {dest_file_to_be_deleted}")

        # 2. Find CSV files in the source directory
        csv_files = sorted(source_dir.glob(f"{lang_code}.csv"))

        if not csv_files:
            print(f"No CSV files found for {lang_code}: {source_dir}")
            continue

        # Ensure the destination directory exists
        dest_dir.mkdir(parents=True, exist_ok=True)

        # 3. Copy the first CSV file as clean1.csv
        shutil.copy2(csv_files[0], dest_file_to_created)
        print(f"Copied: {csv_files[0]} -> {dest_file_to_created}")

move_data()
```
