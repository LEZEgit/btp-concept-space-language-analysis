## Script to drop unnecessary columns from csv

```python
import pandas as pd


prefix = "../data/langs/"
# df_en_fr = pd.read_csv(f'{prefix}{input_lang}/clean.csv').reindex() # <-- taken from Translation.ipynb 
# 1. Load the CSV file into a DataFrame
df = pd.read_csv(f'{prefix}en/clean.csv').reindex()

# 2. Specify the columns you want to drop in a list
columns_to_drop = ['blank_prompt_original', 'blank_prompt_translation', 'blank_prompt_translation_masked', 'error', 'lang', 'word_translation']

# 3. Drop the columns (axis=1 specifies columns rather than rows)
df_cleaned = df.drop(columns = columns_to_drop, axis=1)

# X. Rename the columns
columns_renaming = {'word_original':'en_words'}
df_cleaned.rename(columns = columns_renaming, inplace = True);

# 4. Save the modified data to a new CSV file (excluding the row index)
df_cleaned.to_csv('./en_compressed.csv', index=False)

print("Columns dropped successfully!")
```
